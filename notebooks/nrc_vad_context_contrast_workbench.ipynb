# Context contrast with NRC VAD v2.1

Evaluation of the same 20 English pairs used in the BERT audit, with lexical coverage, per-pair differences, and magnitude comparisons. The lexicon must be available in the local cache: run `uv run python scripts/audit_nrc_vad_context.py --download` from the project root before the first execution.

More variation does not demonstrate greater validity. This evaluation does not validate Portuguese and preserves the default STDI scorer.


```python
from pathlib import Path

import pandas as pd
from IPython.display import Markdown, display

from misinformation_simulation.audits.nrc_vad_context import run_nrc_context_audit

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "pyproject.toml").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
INPUT_PATH = PROJECT_ROOT / "data/synthetic/vad_context_contrast_news.csv"
LEXICON_PATH = PROJECT_ROOT / ".cache/nrc_vad/NRC-VAD-Lexicon-v2.1.zip"
BASELINE_PATH = (
    PROJECT_ROOT / "output/audit/VADContextContrastAudit/vad_context_contrast_scored.csv"
)
OUTPUT_DIR = PROJECT_ROOT / "output/audit/NRCVADContextContrastAudit"
```

## Run the audit

Matches are aggregated by their mean across occurrences, prioritizing the longest expression. Both native values and values converted to the 1-5 scale are exported.


```python
manifest = run_nrc_context_audit(INPUT_PATH, LEXICON_PATH, OUTPUT_DIR, baseline_path=BASELINE_PATH)
print(f"Documents: {manifest['documents']}; pairs: {manifest['pairs']}")
print(f"Token coverage: {manifest['weighted_token_coverage']:.1%}")
```

## Compare dimensions and inspect the pairs


```python
comparison = pd.read_csv(OUTPUT_DIR / "vad_context_contrast_model_comparison.csv")
pair_comparison = pd.read_csv(OUTPUT_DIR / "vad_context_contrast_model_pair_comparison.csv")
scored = pd.read_csv(OUTPUT_DIR / "vad_context_contrast_scored.csv")
display(comparison)
display(pair_comparison.head(10))
display(scored.sort_values("nrc_token_coverage").head(10))
```
